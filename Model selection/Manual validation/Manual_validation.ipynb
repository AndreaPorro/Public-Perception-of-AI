```python
import joblib
import numpy as np
import pandas as pd
from transformers import BertTokenizer, BertModel
import torch
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# ===============================
# 1. CARICA IL MODELLO SALVATO
# ===============================

print("Caricamento modello...")

# Carica il classificatore
classifier = joblib.load('sentiment_classifier.pkl')

# Carica i metadati
model_info = joblib.load('model_info.pkl')

print(f"✓ Modello caricato: {model_info['model_name']}")
print(f"✓ Accuracy training: {model_info['accuracy']:.4f}")


# ===============================
# 2. CARICA BERT PER EMBEDDINGS
# ===============================

print("\nCaricamento BERT...")

bert_model_name = model_info['bert_model']
tokenizer = BertTokenizer.from_pretrained(bert_model_name)
bert_model = BertModel.from_pretrained(bert_model_name)
bert_model.eval()

# Usa GPU se disponibile
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
bert_model.to(device)

print(f"✓ BERT caricato: {bert_model_name}")
print(f"✓ Device: {device}")


# ===============================
# 3. FUNZIONE PER ESTRARRE EMBEDDINGS
# ===============================

def get_embeddings(texts, batch_size=32):
    """Estrae embeddings BERT per una lista di testi"""
    embeddings = []
    
    print(f"Estraendo embeddings per {len(texts)} testi...")
    
    for i in range(0, len(texts), batch_size):
        if i % (batch_size * 50) == 0:
            print(f"Progress: {i}/{len(texts)} ({i/len(texts)*100:.1f}%)")
        
        batch_texts = texts[i:i+batch_size]
        
        encoded = tokenizer(
            batch_texts,
            padding=True,
            truncation=True,
            max_length=512,
            return_tensors='pt'
        )
        
        encoded = {k: v.to(device) for k, v in encoded.items()}
        
        with torch.no_grad():
            outputs = bert_model(**encoded)
            cls_embeddings = outputs.last_hidden_state[:, 0, :].cpu().numpy()
            embeddings.append(cls_embeddings)
    
    return np.vstack(embeddings)


# ===============================
# 4. CARICA IL DATASET df_merged
# ===============================

# Assumendo che df_merged sia già caricato
# df_merged = pd.read_csv('tuo_file.csv')

print(f"\n✓ Dataset caricato: {len(df_merged)} osservazioni")
print(f"Colonne: {list(df_merged.columns)}")

# Verifica che le colonne esistano
assert 'body' in df_merged.columns, "Colonna 'body' non trovata!"
assert 'sentiment' in df_merged.columns, "Colonna 'sentiment' non trovata!"


# ===============================
# 5. FAI LE PREDIZIONI
# ===============================

# Estrai i testi dalla colonna 'body'
texts = df_merged['body'].tolist()

# Estrai embeddings
embeddings = get_embeddings(texts)

# Predici
predictions = classifier.predict(embeddings)

# Converti in label testuali
label_map = model_info['label_mapping']
sentiments = [label_map[pred] for pred in predictions]

print(f"\n✓ Predizioni completate!")


# ===============================
# 6. AGGIUNGI PREDIZIONI AL DATAFRAME
# ===============================

df_merged['predicted_label'] = predictions
df_merged['predicted_sentiment'] = sentiments

# Se il modello supporta probabilità
if hasattr(classifier, 'predict_proba'):
    probas = classifier.predict_proba(embeddings)
    df_merged['prob_negativo'] = probas[:, 0]
    df_merged['prob_neutro'] = probas[:, 1]
    df_merged['prob_positivo'] = probas[:, 2]


# ===============================
# 7. CALCOLA ACCURACY
# ===============================

print("\n" + "="*70)
print("VALUTAZIONE PRESTAZIONI")
print("="*70)

# Ottieni le label vere dalla colonna 'sentiment'
y_true = df_merged['sentiment'].values
y_pred = df_merged['predicted_label'].values

# Calcola accuracy
accuracy = accuracy_score(y_true, y_pred)
print(f"\nAccuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")

# Classification Report dettagliato
print("\nClassification Report:")
print(classification_report(y_true, y_pred, 
                          target_names=['Negativo (-1)', 'Neutro (0)', 'Positivo (1)']))

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Negativo', 'Neutro', 'Positivo'],
            yticklabels=['Negativo', 'Neutro', 'Positivo'])
plt.title(f'Confusion Matrix - Accuracy: {accuracy:.4f}')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.tight_layout()
plt.show()


# ===============================
# 8. ANALISI ERRORI
# ===============================

print("\n" + "="*70)
print("ANALISI ERRORI")
print("="*70)

# Identifica predizioni corrette e sbagliate
df_merged['correct'] = (df_merged['sentiment'] == df_merged['predicted_label'])

correct_count = df_merged['correct'].sum()
incorrect_count = len(df_merged) - correct_count

print(f"\nPredizioni corrette: {correct_count} ({correct_count/len(df_merged)*100:.2f}%)")
print(f"Predizioni sbagliate: {incorrect_count} ({incorrect_count/len(df_merged)*100:.2f}%)")

# Mostra alcuni esempi di errori
print("\n" + "-"*70)
print("ESEMPI DI PREDIZIONI SBAGLIATE")
print("-"*70)

errors = df_merged[~df_merged['correct']].head(5)
for idx, row in errors.iterrows():
    print(f"\nTesto: {row['body'][:100]}...")
    print(f"Vero: {label_map[row['sentiment']]} ({row['sentiment']})")
    print(f"Predetto: {row['predicted_sentiment']} ({row['predicted_label']})")


# ===============================
# 9. STATISTICHE PREDIZIONI
# ===============================

print("\n" + "="*70)
print("DISTRIBUZIONE PREDIZIONI vs REALI")
print("="*70)

print("\nDistribuzione REALE:")
print(df_merged['sentiment'].value_counts().sort_index())

print("\nDistribuzione PREDETTA:")
print(df_merged['predicted_label'].value_counts().sort_index())

# Grafico comparativo
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Distribuzione reale
real_counts = df_merged['sentiment'].value_counts().sort_index()
axes[0].bar(['Negativo', 'Neutro', 'Positivo'], real_counts.values, color='steelblue', alpha=0.7)
axes[0].set_title('Distribuzione Reale')
axes[0].set_ylabel('Conteggio')

# Distribuzione predetta
pred_counts = df_merged['predicted_label'].value_counts().sort_index()
axes[1].bar(['Negativo', 'Neutro', 'Positivo'], pred_counts.values, color='coral', alpha=0.7)
axes[1].set_title('Distribuzione Predetta')
axes[1].set_ylabel('Conteggio')

plt.tight_layout()
plt.show()


# ===============================
# 10. SALVA I RISULTATI
# ===============================

df_merged.to_csv('df_merged_con_predizioni.csv', index=False)
print(f"\n✓ Risultati salvati in 'df_merged_con_predizioni.csv'")
```
